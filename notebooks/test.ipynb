```python
from stalk.star import Star

import matplotlib.pyplot as plt

sun = Star()

xx, yy, intensity = sun.intensity_grid(1000)

plt.imshow(intensity, cmap ='hot', extent=[-1,1,-1,1])

plt.colorbar(label = "I(mu)/I(1)")

plt.title("Solar disk plot with Limb Darkening (quadratic law)")
```
